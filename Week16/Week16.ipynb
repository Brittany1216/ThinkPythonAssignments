```python
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS
# 1. Define category
def get_single_category():
#    - User must enter type of category
    category = input('Enter category: ')
    return category

expenses = []

print('**** Expense Entry System ****')
print("Enter your expense details below. Type 'done' as the category to stop.")

while True:
    category = input("Category (e.g., Groceries, Rent) or 'done' to finish: ")
    if category.lower() == 'done':
        break
# 2. Define vendor
    vendor = input('Vendor: ')
#    - User must enter type of store
# 3. Define amount
#    - User must enter amount
    try:
        amount = float(input('Amount: '))
    except ValueError:
        print('Invalid amount. Please enter a number.')
        continue
# 4. Define date
#    - User must enter a date
    date = input('Date (YYYY-MM-DD): ')

    record = {
        'Category': category,
        'Vendor': vendor,
        'Amount': amount,
        'Date': date
    }

    expenses.append(record)

    print('[TEST] Current Raw Expense List:')
    print(expenses)

print('Final list of expenses captured:')
for item in expenses:
    print(item)


# PROGRAM LOGIC
```
